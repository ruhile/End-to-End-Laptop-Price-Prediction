# 📅 Week 18 – Day 2: Data Cleaning & Preprocessing

This notebook focuses on handling missing values, removing duplicate rows, cleaning text columns, standardizing string representations, and exporting a clean dataset (`laptops_clean.csv`).

## 1️⃣ Import Libraries

```python
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer
import os
```

## 2️⃣ Load Dataset

```python
path = "../data/raw/laptops.csv"
if not os.path.exists(path):
    path = "data/raw/laptops.csv"

df = pd.read_csv(path)
print(f"Loaded dataset from: {path}")
df.head()
```

## 3️⃣ Check Missing Values & Imputation

```python
print("Missing values before imputation:")
print(df.isnull().sum().sort_values(ascending=False))

# Handle numerical missing values
num_cols = df.select_dtypes(include=np.number).columns
if df[num_cols].isnull().sum().sum() > 0:
    imputer_num = SimpleImputer(strategy="median")
    df[num_cols] = imputer_num.fit_transform(df[num_cols])

# Handle categorical missing values
cat_cols = df.select_dtypes(include="object").columns
if df[cat_cols].isnull().sum().sum() > 0:
    imputer_cat = SimpleImputer(strategy="most_frequent")
    df[cat_cols] = imputer_cat.fit_transform(df[cat_cols])
```

## 4️⃣ Remove Duplicates

```python
print("Number of duplicates:", df.duplicated().sum())
df = df.drop_duplicates()
print("Shape after duplicate removal:", df.shape)
```

## 5️⃣ Check Data Types

```python
df.dtypes
```

## 6️⃣ Clean Text Columns

```python
# Strip extra whitespaces from text columns
object_cols = df.select_dtypes(include="object").columns
for col in object_cols:
    df[col] = df[col].astype(str).str.strip()

df.head()
```

## 7️⃣ Standardize Text

```python
# Standardize Brand column
if "Brand" in df.columns:
    df["Brand"] = df["Brand"].str.title()

df["Brand"].value_counts()
```

## 8️⃣ Verify Missing Values Again

```python
print("Missing values after cleaning:")
print(df.isnull().sum())
```

## 9️⃣ Save Clean Dataset

```python
out_path = "../data/processed/laptops_clean.csv"
if not os.path.exists("../data/processed"):
    out_path = "data/processed/laptops_clean.csv"

df.to_csv(out_path, index=False)
print(f"Clean dataset successfully saved to {out_path}")
```
