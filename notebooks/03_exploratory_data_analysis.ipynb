# 📅 Week 18 – Day 3: Exploratory Data Analysis (EDA)

This notebook presents an exploratory analysis of laptop data, examining price distributions, brand popularity, price comparisons by brand and status, top expensive models, correlations, outliers, and key business insights.

## 1️⃣ Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

sns.set_style("whitegrid")
pd.set_option('display.max_columns', None)
```

## 2️⃣ Load Clean Dataset

```python
path = "../data/processed/laptops_clean.csv"
if not os.path.exists(path):
    path = "data/processed/laptops_clean.csv"

df = pd.read_csv(path)
price_col = 'Final Price' if 'Final Price' in df.columns else 'Price'
print(f"Loaded dataset from {path}. Target column: '{price_col}'")
```

## 3️⃣ Dataset Overview

```python
print("Shape:", df.shape)
display(df.head())
df.info()
```

## 4️⃣ Price Distribution

```python
plt.figure(figsize=(10,6))
sns.histplot(df[price_col], bins=30, kde=True, color="skyblue")
plt.title("Distribution of Laptop Prices")
plt.xlabel("Price (€)")
plt.ylabel("Count")
plt.show()
```

## 5️⃣ Brand Distribution

```python
plt.figure(figsize=(12,6))
sns.countplot(y=df["Brand"], order=df["Brand"].value_counts().index, palette="viridis")
plt.title("Number of Laptops by Brand")
plt.xlabel("Count")
plt.ylabel("Brand")
plt.show()
```

## 6️⃣ Average Price by Brand

```python
brand_price = df.groupby("Brand")[price_col].mean().sort_values()

plt.figure(figsize=(12,6))
sns.barplot(x=brand_price.values, y=brand_price.index, palette="magma")
plt.title("Average Laptop Price by Brand (€)")
plt.xlabel("Average Price (€)")
plt.ylabel("Brand")
plt.show()
```

## 7️⃣ Status vs Price

```python
if "Status" in df.columns:
    plt.figure(figsize=(8,5))
    sns.boxplot(x="Status", y=price_col, data=df, palette="Set2")
    plt.title("Price Distribution by Status (New vs Refurbished)")
    plt.show()
```

## 8️⃣ Top 10 Most Expensive Models

```python
top10 = df.nlargest(10, price_col)
cols = [c for c in ["Brand", "Model", price_col] if c in df.columns]
display(top10[cols])
```

## 9️⃣ Correlation Heatmap

```python
numeric = df.select_dtypes(include=np.number)
plt.figure(figsize=(8,6))
sns.heatmap(numeric.corr(), annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap of Numeric Features")
plt.show()
```

## 🔟 Outlier Detection

```python
plt.figure(figsize=(8,5))
sns.boxplot(x=df[price_col], color="coral")
plt.title("Price Outlier Boxplot")
plt.xlabel("Price (€)")
plt.show()
```

## 1️⃣2️⃣ Key Insights

- **Price Skewness:** Laptop prices are strongly right-skewed. The median price is around €1,032, while luxury/gaming models reach up to €7,150.
- **Brand Dominance & Premium Pricing:** Asus, Lenovo, and HP have the largest volume of laptops listed. Premium workstation/gaming brands like MSI, Razer, and Apple command much higher average prices.
- **Status Difference:** New laptops generally maintain higher median prices compared to refurbished units.
- **Outliers:** Extremely high prices correspond to specialized workstation and high-end gaming laptops (e.g. RTX 4090 / high RAM builds) and should be preserved for model training.
- **Feature Needs:** Currently, numerical correlations are limited because RAM, Storage, and Screen are stored as numerical or text fields that need feature engineering in Day 4.
